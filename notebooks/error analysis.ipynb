# Explicitly paired error analysis

Duplicate IDs, missing pairs and non-finite values fail rather than silently becoming zeros.

```python
from pathlib import Path
import os
import sys
HERE = Path.cwd().resolve()
MODULE_DIR = next(p for p in [HERE, HERE/'notebooks', HERE.parent/'notebooks']
                  if (p/'benchmark_workflows.py').is_file())
sys.path.insert(0,str(MODULE_DIR))
import benchmark_workflows as workflows
REPOSITORY = workflows.repo_root()

def compare_tables(reference_csv,prediction_csv,reference_column,prediction_column):
    def index(rows):
        out={}
        for row in rows:
            key=row['ID']
            if key in out:raise ValueError(f'Duplicate ID: {key}')
            out[key]=row
        return out
    a=index(workflows.table(reference_csv));b=index(workflows.table(prediction_csv))
    if set(a)!=set(b):raise ValueError('Reference and predicted ID sets differ.')
    keys=sorted(a)
    return workflows.metrics([float(a[k][reference_column]) for k in keys],
                              [float(b[k][prediction_column]) for k in keys])
RUN_ANALYSIS=False
if RUN_ANALYSIS:
    print(compare_tables(REPOSITORY/'inputs/reference.csv',
                         REPOSITORY/'results/predictions.csv','reference','prediction'))
```
