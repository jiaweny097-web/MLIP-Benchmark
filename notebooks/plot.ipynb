# Selected-table plotting

Plot selected metrics from consolidated PKLs. Final I1 is pending.

```python
from pathlib import Path
import os
import sys
HERE = Path.cwd().resolve()
MODULE_DIR = next(p for p in [HERE, HERE/'notebooks', HERE.parent/'notebooks']
                  if (p/'benchmark_workflows.py').is_file())
sys.path.insert(0,str(MODULE_DIR))
import benchmark_workflows as workflows
REPOSITORY = workflows.repo_root()

import numpy as np
import matplotlib.pyplot as plt
def plot_metric_table(path,column='MAE'):
    rows=[r for r in workflows.table(path) if r.get('task')=='vacancy']
    values=np.array([float(r[column]) for r in rows])
    if not np.isfinite(values).all():raise ValueError('Non-finite metric values.')
    fig,ax=plt.subplots(figsize=(8,4))
    ax.bar([r['model'] for r in rows],values)
    ax.set_ylabel(column);ax.tick_params(axis='x',rotation=25);fig.tight_layout()
    return fig
PLOT_METRICS=False
if PLOT_METRICS:
    fig=plot_metric_table(REPOSITORY/'results/consolidated/vacancy_surface_metrics.csv')
    fig.suptitle('Vacancy metrics from consolidated records')
    output=REPOSITORY/'results/vacancy_metrics.png'
    output.parent.mkdir(parents=True,exist_ok=True)
    fig.savefig(output,dpi=300,bbox_inches='tight')
```
